```python
import requests
r = requests.post("https://1.1.1.1/dns-query",
                  headers={"accept": "application/dns-message",
                           "content-type": "application/dns-message",
                           "host": "cloudflare-dns.com"},
                  data=b"\x00\x00", verify=True, timeout=5)
print(r.status_code)
```
